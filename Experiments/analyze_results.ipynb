```python
data_alloc = gem_data[(gem_data['Good A'] == '1') & (gem_data['Good B'] == '2')]
data_alloc['Person 1 money'] = pd.to_numeric(data_alloc['Person 1 money'], errors='coerce')
data_alloc['Person 2 money'] = pd.to_numeric(data_alloc['Person 2 money'], errors='coerce')
data_alloc = data_alloc.dropna()
data_alloc['diffs'] = data_alloc['Person 2 money'] - data_alloc['Person 1 money']
data_alloc['sums'] = data_alloc['Person 2 money'] + data_alloc['Person 1 money']
data_alloc[data_alloc['diffs'] == 20].shape
data_alloc[data_alloc['sums'] == 50].shape
# data_alloc[(data_alloc['diffs'] >= 20) & (data_alloc['diffs'] <= 30)].shape
# data_alloc.shape
```

```python
data_alloc = gem_data[(gem_data['Good A'] == '2') & (gem_data['Good B'] == '1')]
data_alloc['Person 1 money'] = pd.to_numeric(data_alloc['Person 1 money'], errors='coerce')
data_alloc['Person 2 money'] = pd.to_numeric(data_alloc['Person 2 money'], errors='coerce')
data_alloc = data_alloc.dropna()
data_alloc['diffs'] = data_alloc['Person 1 money'] - data_alloc['Person 2 money']
data_alloc[data_alloc['diffs'] == 20].shape
# data_alloc.shape
```

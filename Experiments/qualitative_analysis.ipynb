```python
import pandas as pd
import os
import pathlib
import textwrap
from IPython.display import display
from IPython.display import Markdown
from components import PromptGenerator, LLM
import time
import json
```

```python
def to_markdown(text):
  text = text.replace('•', '  *')
  return Markdown(textwrap.indent(text, '> ', predicate=lambda _: True))
```

```python
def get_response(family, model, question, valuation, restriction, template, justification):
    llm = LLM(family=family, model=model)
    pg = PromptGenerator(question=question,valuation_type=valuation,restriction=restriction, template=template, justification=justification, llm_type='chatgpt')
    num_agents, num_goods = pg.set_text_values()
    prompt1 = pg.generate_prompt1()

    print(prompt1)

    df_dict = {
        'Text': [],
    }

    for g in range(num_goods):
        df_dict[f'Good {chr(65+g)}'] = []

    df_dict['Text'] = llm.makeLLMRequest(prompt1)
    prompt2 = pg.generate_prompt2(df_dict['Text'])
    print(prompt2)

    try:
        alloc = llm.makeLLMRequest(prompt2)
    except:
        print('LLM request failed.')
        time.sleep(5)

    bopen = alloc.find('{')
    bclose = alloc.find('}')

    json_obj = json.loads(alloc[bopen:bclose+1])
    for g in range(num_goods):
        df_dict[f'Good {chr(65+g)}'].append(json_obj[f'Good {chr(65+g)}'].split()[-1])
    if 'money' in pg.valuations:
        for a in list(pg.valuations['valuation'].keys()):
            df_dict[f'{pg.person_names[a]} money'] = json_obj[f'{pg.person_names[a]} money']

    print(df_dict)
```

```python
def get_single_response(auth_file, family, model, question, valuation, restriction, template):
    llm = LLM(auth_file=auth_file, family=family, model=model)
    pg = PromptGenerator(question=question,valuation_type=valuation,restriction=restriction, template=template,llm_type=family)
    num_agents, num_goods = pg.set_text_values()
    prompt1 = pg.generate_prompt1()

    print(prompt1)

    df_dict = {
        'Text': [],
    }

    for g in range(num_goods):
        df_dict[f'Good {chr(65+g)}'] = []

    df_dict['Text'] = llm.makeLLMRequest(prompt1)
    prompt2 = pg.generate_prompt2(df_dict['Text'])
    # print(prompt2)

    try:
        alloc = llm.makeLLMRequest(prompt2)
    except:
        print('LLM request failed.')
        time.sleep(5)

    bopen = alloc.find('{')
    bclose = alloc.find('}')

    json_obj = json.loads(alloc[bopen:bclose+1])
    
    for g in range(num_goods):
        df_dict[f'Good {chr(65+g)}'].append(json_obj[f'Good {chr(65+g)}'].split()[-1])
    if 'money' in pg.valuations:
        for a in list(pg.valuations['valuation'].keys()):
            df_dict[f'{pg.person_names[a]} money'] = json_obj[f'{pg.person_names[a]} money']

    return df_dict
```
